# TinyWorlds playground

This read-only notebook explores the deterministic TinyWorlds knowledge graph and the completed calibration evidence. The calibration stopped because no facts-axis configuration passed every scientific gate, so no calibration profile was locked and Phase 5 was not launched.

Only saved validation evidence is shown here. The held-out calibration test remained unopened, and this notebook does not train, tune, or score a model. Candidate comparisons also carry a lexical-prior confound: a frozen language model can prefer one nonce answer string over another without learning the fictional relation. The 64/128 counts are paired renderings and candidate rotations of a very small number of semantic plans, not independent facts or stories.

In this benchmark, **one-hop** means exactly one Horn-rule application in the symbolic proof. It does not mean one neural-network layer or one edge in the continual-learning task graph.

For the interactive controls and local language-model assets, install the optional dependencies from the repository root with `pip install -e '.[notebook,lm]'`.

```python
from IPython.display import HTML, display

from apm.interactive.tinyworlds import (
    QueryKind,
    exact_kg_summary,
    generate_tinyworlds_demo,
    inspect_query,
    load_tinyworlds_lab,
    status_html,
    tinyworlds_playground,
)
```

```python
RESULT_DIRECTORY = None
PUBLIC_SEED = 0
```

```python
lab = load_tinyworlds_lab(result_directory=RESULT_DIRECTORY)
demo = generate_tinyworlds_demo(lab, public_seed=PUBLIC_SEED)
```

## Calibration status

The status panel distinguishes the mechanically valid benchmark from the failed scientific calibration gates.

```python
display(HTML(status_html(lab)))
```

## Unified playground

Use the controls to move between symbolic queries, cue conditions, proofs, candidate roles, and the saved validation diagnostics without changing the benchmark.

```python
display(tinyworlds_playground(lab, demo))
```

## Direct API: inspect a one-hop proof

This selects a query whose canonical symbolic proof applies one Horn rule exactly once.

```python
display(inspect_query(demo, QueryKind.ONE_HOP))
```

## Direct API: exact knowledge-graph ceiling

The exact executor checks symbolic answers and canonical proofs. Its ceiling is separate from learned-model candidate accuracy.

```python
display(exact_kg_summary(demo))
```
